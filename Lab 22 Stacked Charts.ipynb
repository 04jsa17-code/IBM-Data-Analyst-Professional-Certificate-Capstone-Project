### Load the data

### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
##Write your code here
age_map = {
    'Under 18 years old': 16, '18-24 years old': 21,
    '25-34 years old': 30, '35-44 years old': 40,
    '45-54 years old': 50, '55-64 years old': 60,
    '65 years or older': 70
}

df['JobSatPoints_6'] = pd.to_numeric(df['JobSatPoints_6'], errors='coerce')
df['JobSatPoints_7'] = pd.to_numeric(df['JobSatPoints_7'], errors='coerce')
df['Age_numeric'] = df['Age'].map(age_map)

median_sat = df.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median()

median_sat.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Median JobSatPoints_6 and JobSatPoints_7 by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Median Job Satisfaction')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
##Write your code here
top5_emp = df['Employment'].value_counts().head(5).index
df_emp = df[df['Employment'].isin(top5_emp)]

median_emp = df_emp.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()

median_emp.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('JobSatPoints_6 and JobSatPoints_7 by Employment Status')
plt.xlabel('Employment Type')
plt.ylabel('Median Job Satisfaction')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
##Write your code here
df['ConvertedCompYearly'] = pd.to_numeric(df['ConvertedCompYearly'], errors='coerce')
df_age = df[df['Age'] == '25-34 years old'].copy()
df_age = df_age[df_age['ConvertedCompYearly'] < 1_000_000]

median_data = pd.DataFrame({
    'ConvertedCompYearly': [df_age['ConvertedCompYearly'].median()],
    'JobSatPoints_6': [df_age['JobSatPoints_6'].median()]
}, index=['Age 30-35'])

median_data.plot(kind='bar', stacked=True, figsize=(8, 6))
plt.title('Compensation and Job Satisfaction for Age Group 30-35')
plt.ylabel('Median Value')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
##Write your code here
df_plot = df[df['ConvertedCompYearly'] < 1_000_000].copy()
median_age = df_plot.groupby('Age')[['ConvertedCompYearly', 'JobSatPoints_6']].median()

median_age.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Median Compensation and Job Satisfaction Across Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Median Value')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
##Write your code here
df_db = df[['DatabaseWantToWorkWith', 'Age']].dropna()
df_db = df_db.assign(DatabaseWantToWorkWith=df_db['DatabaseWantToWorkWith'].str.split(';')).explode('DatabaseWantToWorkWith')
df_db['DatabaseWantToWorkWith'] = df_db['DatabaseWantToWorkWith'].str.strip()

top5_db = df_db['DatabaseWantToWorkWith'].value_counts().head(5).index
df_db = df_db[df_db['DatabaseWantToWorkWith'].isin(top5_db)]

db_age = df_db.groupby(['Age', 'DatabaseWantToWorkWith']).size().unstack(fill_value=0)

db_age.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Preferred Databases by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(loc='upper right', fontsize=7)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python

##Write your code here
df_es = df[['Employment', 'JobSatPoints_6']].dropna()
df_es['JobSatPoints_6'] = pd.to_numeric(df_es['JobSatPoints_6'], errors='coerce').dropna()
df_es = df_es.dropna()
df_es['JobSatPoints_6'] = df_es['JobSatPoints_6'].astype(int)

top5_emp = df_es['Employment'].value_counts().head(5).index
df_es = df_es[df_es['Employment'].isin(top5_emp)]

emp_sat = df_es.groupby(['JobSatPoints_6', 'Employment']).size().unstack(fill_value=0)

emp_sat.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Employment Type by Job Satisfaction Level')
plt.xlabel('Job Satisfaction')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.legend(loc='upper right', fontsize=7)
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
##Write your code here
df_lang = df[['LanguageAdmired', 'Age']].dropna()
df_lang = df_lang.assign(LanguageAdmired=df_lang['LanguageAdmired'].str.split(';')).explode('LanguageAdmired')
df_lang['LanguageAdmired'] = df_lang['LanguageAdmired'].str.strip()

top5_lang = df_lang['LanguageAdmired'].value_counts().head(5).index
df_lang = df_lang[df_lang['LanguageAdmired'].isin(top5_lang)]

lang_age = df_lang.groupby(['Age', 'LanguageAdmired']).size().unstack(fill_value=0)

lang_age.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Preferred Programming Languages by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(loc='upper right', fontsize=7)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
##Write your code here
df_plat = df[['PlatformAdmired', 'Employment']].dropna()
df_plat = df_plat.assign(PlatformAdmired=df_plat['PlatformAdmired'].str.split(';')).explode('PlatformAdmired')
df_plat['PlatformAdmired'] = df_plat['PlatformAdmired'].str.strip()

top5_plat = df_plat['PlatformAdmired'].value_counts().head(5).index
top5_emp = df_plat['Employment'].value_counts().head(5).index
df_plat = df_plat[df_plat['PlatformAdmired'].isin(top5_plat) & df_plat['Employment'].isin(top5_emp)]

plat_emp = df_plat.groupby(['Employment', 'PlatformAdmired']).size().unstack(fill_value=0)

plat_emp.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('Technology Adoption by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(loc='upper right', fontsize=7)
plt.tight_layout()
plt.show()
```
